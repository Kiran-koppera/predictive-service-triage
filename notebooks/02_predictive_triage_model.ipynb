# Predictive Triage Model

Trains and evaluates the triage-priority classifier from src/model.py on the engineered features from src/feature_engineering.py. Evidence for "Predictive Triage & Machine Learning" in docs/00_role_alignment.md.

```python
import pandas as pd
from src import feature_engineering, model

df = pd.read_parquet('../data/processed/incident_triage_features.parquet')
df = feature_engineering.add_recency_features(df)
df = feature_engineering.add_severity_label(df)
df.head()
```

```python
clf, report = model.train(df)
print(report)
```

```python
import pandas as pd

importances = pd.Series(clf.feature_importances_, index=model.FEATURE_COLS).sort_values(ascending=False)
importances.plot(kind='barh', title='Feature importance')
```
