```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet('../data/processed/incident_triage_features.parquet')
df.head()
```

```python
df.groupby('model')['event_id'].count().sort_values(ascending=False).plot(kind='bar', title='Incidents by model')
plt.tight_layout()
plt.show()
```
